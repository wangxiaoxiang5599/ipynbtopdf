# Housing price analysis

A quick look at the **Ames** dataset. We fit a simple model and check the residuals.

The loss we minimise is $\mathcal{L} = \frac{1}{n}\sum_i (y_i - \hat{y}_i)^2$, and the closed form solution is

$$\hat{\beta} = (X^\top X)^{-1} X^\top y$$

Steps:

1. Load the data
2. Fit the model
3. Inspect residuals
